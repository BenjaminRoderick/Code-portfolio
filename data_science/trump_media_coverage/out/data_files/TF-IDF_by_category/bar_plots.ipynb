```python
import matplotlib.pyplot as plt
import numpy as np

def plot_csv(filename, title):
    word_list = []
    value_list = []
    with open(filename, 'r') as f:
        for i in range(10):
            line_tuple = f.readline().rstrip().split(',')
            word_list.append(line_tuple[0])
            value_list.append(float(line_tuple[1]))

    terms = np.array(word_list)
    scores = np.array(value_list)

    fig = plt.figure(figsize=(15,7))
    plt.bar(terms, scores, width=0.6)
    plt.xlabel('Term')
    plt.ylabel('TF-IDF score')
    plt.title(title)
    plt.show()
```

```python
plot_csv('climate_TFIDF.csv', 'Climate')
```

```python
plot_csv('economy_TFIDF.csv', 'Economy')
```

```python
plot_csv('election_TFIDF.csv', 'Election')
```
