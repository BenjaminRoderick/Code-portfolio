# HOW TO: run this project

The first step is to upload all your images into the "images" directory, then input your desired parameters into a .csv or .json file (a template is included for your convenience) that must be located in the same directory as the scripts.

```python
pip install -r requirements.txt
```

Now that the setup is done, we can move on to running the file. 

## For Unix users
You can do this from the command line using:

```python
python deblur_images.py [problem] [algorithm] [image_filename] [blur_type] [parameters_filename]
```

## For Windows
Slightly more complicated, I prefer to run the program once with no arguments to get your interpreter to call it for you, then copy-paste the call from the command line and add the arguments to it before running the program again with the real inputs you want.

### Additionally, there are two optional flags you can add at the end:

-g : performs a grid search to find the optimal values for t and gamma (very slow!)

-s : saves the output of the script to a file

## The meaning for different parameters in the default_parameters.csv file
- kernel_size: length of one side of the square blurring kernel you want to generate
- resize_factor: the image will be down (or up) scaled by this factor (try 0.5 for a significant speedup!)
- noise_density: the amount of noise to add to the blurred image
- noise_type: the type of noise to add to the blurred image (options are gaussian or s&p)
- sigma: strength of gaussian blurring kernel
- angle: angle of motion blur kernel
- alpha: strength of motion blur kernel
- mu: cost function scaling parameter
- rho: relaxation parameter
- t: primal problem step size
- s: dual problem step size
- gamma: de-noising strength parameter
- max_iter: maximum number of iterations before termination (less if stopping criterion is met)
